### Model integration with  OPENROUTER GOOGLE GEMINI AND GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-2.5-flash-lite")
```

#### Invoke the model

```python
response = model.invoke("How do parrots talk?")
response.content
```

#### GROK MODEL INTEGRATION

```python
# I can use init chat model that i used above but lets try new way
import os
from langchain_groq import ChatGroq
model  = ChatGroq(
    model="qwen/qwen3-32b")
response = model.invoke("What is the capital of India?")
print(response.content)
```

#### STREAM AND BATCHING

```python
for chunk in model.stream("What is the capital of India?"):
    print(chunk.content)
```

```python
responses =  model.batch([
    "What is your latest version?",
    "what is the GenAI?"
    "What is the agent and automation"
])
for response in responses:
    print(response.content)
```
