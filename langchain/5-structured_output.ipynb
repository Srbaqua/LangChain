Structured output:
    Models are required to provide their response in a specific format so that subsequent steps can easily extract the relevant information. This format is called structured output.

    -Types of structured output:
    --Pydantic

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"] =os.getenv("GOOGLE_API_KEY")
model = init_chat_model("gemini-2.5-flash-lite")
model
```

PYDANTIC

```python
from pydantic import BaseModel, Field
class Movie(BaseModel):
    title:str=Field(description="Movie title")
    year:int=Field(description = "Movie year")
    rating:float=Field(description="Movie rating")
```

```python
model_with_structure = model.with_structured_output(Movie)
model_with_structure.stream("Give details about movie Harry Potter")
```
